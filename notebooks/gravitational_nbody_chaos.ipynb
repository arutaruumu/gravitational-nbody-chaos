### Abstract

The **Gravitational N-Body Chaos Simulator** is a computational physics framework designed to investigate the dynamical behavior of systems containing multiple mutually interacting bodies under Newtonian gravity. Unlike a simple two-body orbital simulation, an N-body system can exhibit highly nonlinear and chaotic behavior because every body continuously modifies the gravitational field experienced by every other body.

The simulator numerically intgrates the equation of motion for systems containing approcimately **3 to 1000 bodies** and analyzes their dynamical evolution using several physical and numerical diagnostic. These include total energy conservation, angular momentum conservation, trajectory divergence, Lyapunov-like behavior, orbital stability, collision detection, and escape trajectories.

The primary objectibe is not only to visualize gravitational motion, but also to investigate how small changes in initial conditions and system parameters can produce significantly different long-term trajectories.

# Introduction

## Background

The gravitational N-body problem concerns the motion of multiple bodies interacting through their mutual gravitational attraction.

For two bodies, Newtonian gravity produces a well-known analytical solution in which the resulting orbit is generally a conic section: an ellipse, parabola, or hyperbola.

However, when the number of interacting bodies increases beyond two, the mathematical problem becomes considerably more complicated.

For $N$ bodies, every object interacts gravitationally with all other objects. Therefore, the acceleration of body $i$ depends on the positions of all remaining $N - 1$ bodies.
